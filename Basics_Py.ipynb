```python
# ============================================================
# BASE SALARY CLASS
# Purpose:
# Calculate total salary, bonus amount, and bonus percentage.
# Also practice f-string number formatting and __repr__().
# ============================================================


class BaseSalary:

    def __init__(self, base_salary, bonus_rate=.1, symbol="$"):
        # Store the values passed when creating the object.
        self.base_salary = base_salary
        self.bonus_rate = bonus_rate
        self.symbol = symbol

        # Calculate total salary:
        # base salary + bonus percentage of the base salary.
        # Example: 100000 * (1 + 0.10) = 110000
        self.total_salary = base_salary * (1 + bonus_rate)

        # Calculate the actual bonus rate from the salary values.
        # This should give the same rate that was originally provided.
        # Example: (110000 - 100000) / 100000 = 0.10
        self.bonus__rate = (self.total_salary - base_salary) / base_salary

        # Calculate the actual bonus amount.
        # Example: 110000 - 100000 = 10000
        self.bonus = self.total_salary - base_salary


    def __repr__(self):
        # __repr__ should RETURN a string representation of the object.
        # It should not use print().
        #
        # :,.0f means:
        # ,    -> add thousands separators (100000 -> 100,000)
        # .0f  -> display 0 digits after the decimal point
        #
        # Example: 100000.0 -> $100,000
        return f"{self.symbol}{self.base_salary:,.0f}"


    def calculate_salary(self):
        # Display total salary with comma separators
        # and no decimal places.
        #
        # Example: 110000.0 -> $110,000
        print(f"{self.symbol}{self.total_salary:,.0f}")#0f means no number after decimal and ':,' means add separtor ',' to salary as 10000 to 10,000


    def calculate_bonus__rate(self):
        # bonus__rate is stored as a decimal.
        # 0.1 represents 10%.
        #
        # If we want to display it as 10%, we can use:
        # f"{self.bonus__rate:.0%}"
        print(f"{self.symbol}{self.bonus__rate}")


    def calculate_bonus(self):
        # Display the bonus amount with:
        # ,    -> thousands separator
        # .0f  -> no decimal places
        #
        # Example: 10000.0 -> $10,000
        print(f"{self.symbol}{self.bonus:,.0f}")


# Create an object.
# base_salary = 100000
# bonus_rate is not provided, so the default value 0.1 (10%) is used.
# symbol is not provided, so the default "$" is used.
p1 = BaseSalary(100000)

# Total salary = $110,000
p1.calculate_salary()

# Bonus = $10,000
p1.calculate_bonus()

# Bonus rate = 0.1 (which represents 10%)
p1.calculate_bonus__rate()
```

```python
# ============================================================
# APPLYING BONUS RATES TO DIFFERENT JOB SALARIES
# ============================================================

# Create a NumPy array containing different job roles.
jobs = np.array([
    'Data Analyst',
    'Data Scientist',
    'Data Engineer',
    'ML Engineer'
])

# Create the corresponding salary for each job.
# The position of each salary matches the position of its job.
salaries = np.array([20000, 30000, 20000, 35000])

# Create a bonus rate for each job.
# 0.1 = 10%, 0.2 = 20%, etc.
bonus = np.array([.1, .1, .2, .3])

# Calculate the salary after adding the corresponding bonus.
# NumPy performs element-wise multiplication:
#
# salary × (1 + bonus rate)
#
# Example:
# 20000 × 1.1 = 22000
bonused_salaries = salaries * (1 + bonus)
bonused_salaries
```

```python
np.mean(bonused_salaries)
```

```python
# ============================================================
# HANDLING MISSING VALUES WITH NaN
# ============================================================

# Add a new job role.
jobs = np.array([
    'Data Analyst',
    'Data Scientist',
    'Data Engineer',
    'ML Engineer',
    'New Role'
])

# Add a salary for the new role.
# 20000 is being used as a temporary/example salary here.
salaries = np.array([
    20000, 21000, 30000, 35000, 20000
])

# Add bonus rates for each job.
# np.nan represents a missing/unknown value.
#
# np.nan is useful when we don't know the bonus rate
# for the newly added role yet.
bonus = np.array([
    .1, .1, .2, .3, np.nan
])

# Calculate salaries after applying the bonus rates.
# Because the last bonus is np.nan, the corresponding
# bonused salary will also become np.nan.
bonused_salaries = salaries * (1 + bonus)
bonused_salaries
```

```python
np.nanmean(bonused_salaries)
```

```python
import pandas as pd

df = pd.read_csv("ElecSalesData.CSV")


# ---------------- .loc ----------------

# .loc[rows, columns]
# 1st argument → rows
# 2nd argument → columns BY LABEL/NAME

print(df.loc[0:2, "Date":"Quantity"])


# Select row 0 from the 'Date' column
print(df['Date'].loc[0])


# [[]] → use double square brackets when selecting MULTIPLE columns
df[['Date', 'Quantity']].loc[0]


# Multiple conditions in pandas:
# & → AND
# | → OR
# ~ → NOT
#
# IMPORTANT: () around EACH condition are required
# Pandas uses '&' instead of Python's 'and'

df.loc[
    (df['Customer'] == 'Rahul Mehta') & (df['Quantity'] == 5),
    "Date":"Product"
]


# ---------------- .iloc ----------------

# .iloc[rows, columns]
# Uses INTEGER POSITION, NOT column names
#
# 1st argument → row position
# 2nd argument → column position

df.iloc[3, 2]          # row 3, column 2

df.iloc[0:10, 0:2]    # rows 0-9, columns 0-1


# Boolean condition
condition = (df['Customer'] == 'Rahul Mehta') & (df['Quantity'] == 5)

# .iloc can use Boolean array + integer column positions
df.iloc[condition, :]
```
